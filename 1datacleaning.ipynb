```python
df = df.drop_duplicates()
```

```python
df['Country'].unique()
```

```python

df['Gender'].unique()
```

```python
df['Age'] = pd.to_numeric(df['Age'], errors='coerce')
```

```python
df['Purchase_Amount'] = pd.to_numeric(df['Purchase_Amount'], errors='coerce')
```

```python
df
```

```python
df['Gender'] = df['Gender'].str.strip().str.lower()
```

```python

df['Gender'] = df['Gender'].replace({
    'm': 'Male',
    'male': 'Male',
    'female': 'Female',
    'f': 'Female'
})
```

```python
df['Country'] = df['Country'].str.strip().str.lower()

df['Country'] = df['Country'].replace({
    'usa': 'USA',
    'us': 'USA',
    'united states': 'USA',
    'u.k.': 'UK',
    'uk': 'UK',
    'india': 'India',
    'canada': 'Canada'
})
```

```python
#df['Signup_Date'] = pd.to_datetime(df['Signup_Date'],errors='coerce')
```

```python
df['Signup_Date'] = pd.to_datetime(
    df['Signup_Date'],
    errors='coerce',
    infer_datetime_format=True,
    dayfirst=True
)
```

```python
df['Signup_Date'] = df['Signup_Date'].fillna(df['Signup_Date'].median())
```

```python
df['Signup_Date']
```

```python
df['Purchase_Amount']=df['Purchase_Amount'].fillna(df['Purchase_Amount'].median())
```

```python
df['Age']
```

```python
df.loc[(df['Age'] < 0) | (df['Age'] > 100), 'Age'] = None
```

```python
df['Age'] = df['Age'].fillna(df['Age'].median())
```

```python
import re

email_pattern = r'^[A-Za-z0-9._%+-]+@[A-Za-z0-9.-]+\.[A-Za-z]{2,}$'

df['Email_Valid'] = df['Email'].apply(
    lambda x: bool(re.match(email_pattern, str(x))) if pd.notna(x) else False
)


df[df['Email_Valid'] == False][['Email']]
```

```python
df.loc[df['Email_Valid'] == False, 'Email'] = None
```

```python
df['Email'] = df['Email'].fillna('unknown@email.com')
```

```python
df
```

```python
df.info()
```

```python
df['Age'] = df['Age'].astype('Int64')
```

```python
df['Gender'] = df['Gender'].astype('category')
df['Country'] = df['Country'].astype('category')
```

```python
df.to_csv('cleaned_customer_data.csv', index=False)
print("DataFrame 'cleaned_customer_data.csv'")
```
